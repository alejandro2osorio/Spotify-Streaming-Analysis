## Resumen de la exploración inicial

- El dataset contiene 50,000 registros y 33 columnas.
- No se encontraron valores nulos.
- No se encontraron filas duplicadas.
- Los 50,000 `track_id` son únicos.
- El dataset contiene 500 artistas, 20 géneros y 30 países.
- Los datos abarcan los años 2020–2025, con una distribución relativamente equilibrada por año.
- Algunas categorías, especialmente géneros y artistas, presentan una distribución desigual.
- `release_date` está almacenada como texto (`str`) y deberá convertirse a tipo fecha.
- `popularity`, `danceability`, `energy`, `key` y `mode` cumplen los rangos esperados.
- `stream_count` presenta una distribución fuertemente sesgada, con valores extremos altos que deberán analizarse antes de decidir cualquier tratamiento.
